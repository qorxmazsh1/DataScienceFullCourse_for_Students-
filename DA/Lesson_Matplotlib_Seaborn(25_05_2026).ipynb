````python
data = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [1200, 1500, 1350, 1700, 1600, 1900],
    "profit": [300, 420, 360, 500, 460, 620],
    "product_a": [700, 850, 780, 950, 900, 1100],
    "product_b": [500, 650, 570, 750, 700, 800]
})

data
````

## 7. First plot with Matplotlib

Line charts are useful when the order matters, especially for time-based data.

````python
plt.plot(data["month"], data["sales"])
plt.show()
````

## 8. Title and axis labels

A chart should explain what it is showing.

````python
plt.plot(data["month"], data["sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
````

## 9. Figure size

`figsize` controls the width and height of the chart.

````python
plt.figure(figsize=(10, 4))
plt.plot(data["month"], data["sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
````

## 10. Markers

Markers show the exact data points on the line.

````python
plt.figure(figsize=(8, 4))
plt.plot(data["month"], data["sales"], marker=">")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
````

## 11. More than one line

A legend is useful when there are multiple lines.

````python
plt.figure(figsize=(8, 4))
plt.plot(data["month"], data["product_a"], label="Product A" ,marker="o") #, marker="o", label="Product A")
plt.plot(data["month"], data["product_b"], marker="o",  label="Product B") # , label="Product B"
plt.title("Monthly Sales by Product")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()
````

## 12. Bar chart

A bar chart is useful for comparing categories.

````python
plt.figure(figsize=(8, 4))
plt.bar(data["month"], data["sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
````

## 13. Horizontal bar chart

Horizontal bars are useful when category names are long.

````python
plt.figure(figsize=(8, 4))
plt.barh(data["month"], data["sales"])
plt.title("Monthly Sales")
plt.xlabel("Sales")
plt.ylabel("Month")
plt.show()
````

## 14. Scatter plot

A scatter plot shows the relationship between two numeric variables.

````python
plt.figure(figsize=(6, 4))
plt.scatter(data["sales"], data["profit"])
plt.title("Sales and Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
````

## 15. Histogram

A histogram shows the distribution of one numeric variable.

````python
scores = pd.Series([55, 62, 67, 70, 72, 75, 77, 80, 83, 85, 88, 90, 92, 95, 98])
plt.figure(figsize=(7, 4))
plt.hist(scores, bins=10)
plt.title("Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
````

## 16. Pie chart

A pie chart shows parts of a whole. It is best with a small number of categories.

## 17. Grid lines

Grid lines can make values easier to read.

````python
plt.figure(figsize=(8, 4))
plt.plot(data["month"], data["sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
````

## 18. Saving a chart

`savefig()` saves the chart as an image file.

````python
plt.figure(figsize=(8, 4))
plt.plot(data["month"], data["sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.savefig("monthly_sales.png", dpi=150, bbox_inches="tight")
plt.show()
````

## 19. Seaborn basics

Seaborn works well with Pandas DataFrames.

Most Seaborn functions use this style:

```python
sns.function_name(data=df, x="column_name", y="column_name")
```

````python
employees = pd.DataFrame({
    "name": ["Ali", "Leyla", "Murad", "Ayan", "Nigar", "Orkhan", "Sara", "Rauf", "Kamran", "Lala"],
    "department": ["IT", "HR", "IT", "Finance", "Finance", "HR", "IT", "Finance", "Sales", "Sales"],
    "city": ["Baku", "Ganja", "Baku", "Sumgait", "Baku", "Ganja", "Sumgait", "Baku", "Baku", "Ganja"],
    "age": [25, 29, 31, 24, 35, 28, 26, 33, 30, 27],
    "salary": [1800, 2200, 2700, 1950, 3100, 2100, 2400, 2900, 2600, 2300],
    "score": [85, 90, 78, 92, 88, 74, 81, 86, 79, 84],
    "experience": [2, 5, 7, 2, 10, 4, 3, 8, 6, 4]
})

employees
````

## 20. Seaborn scatter plot

`scatterplot()` shows the relationship between two numeric columns.

````python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=employees, x="experience", y="salary")
plt.title("Experience and Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
````

## 21. Scatter plot with hue

A third variable can be shown with `hue`.

````python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=employees, x="experience", y="salary", hue="department")
plt.title("Experience and Salary by Department")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
````

## 22. Seaborn line plot

`lineplot()` is useful for showing change over an ordered variable.

````python
plt.figure(figsize=(8, 4))
sns.lineplot(data=data, x="month", y="sales", marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
````

## 23. Seaborn bar plot

`barplot()` shows a summary value for each category. By default, it shows the mean.

````python
plt.figure(figsize=(7, 4))
sns.barplot(data=employees, x="department", y="salary")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
````

## 24. Count plot

`countplot()` counts how many rows are in each category.

````python
plt.figure(figsize=(7, 4))
sns.countplot(data=employees, x="department")
plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
````

## 25. Box plot

A box plot shows median, spread, and possible unusual values.

````python
plt.figure(figsize=(7, 4))
sns.boxplot(data=employees, x="department", y="salary")
plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()
````

## 26. Histogram with Seaborn

`histplot()` shows the distribution of a numeric column.

````python
plt.figure(figsize=(7, 4))
sns.histplot(data=employees, x="salary", bins=5)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Count")
plt.show()
````

## 27. KDE curve

A KDE curve is a smooth estimate of a distribution.

````python
plt.figure(figsize=(7, 4))
sns.histplot(data=employees, x="salary", kde=True)
plt.title("Salary Distribution with KDE")
plt.xlabel("Salary")
plt.ylabel("Count")
plt.show()
````

## 28. Pair plot

`pairplot()` creates several scatter plots and histograms at once.

````python
sns.pairplot(employees[["age", "salary", "score", "experience"]])
plt.show()
````

## 29. Correlation heatmap

A heatmap can show correlation between numeric columns.

````python
correlation = employees[["age", "salary", "score", "experience"]].corr()
plt.figure(figsize=(7, 5))
sns.heatmap(correlation, annot=True)
plt.title("Correlation Heatmap")
plt.show()
````
