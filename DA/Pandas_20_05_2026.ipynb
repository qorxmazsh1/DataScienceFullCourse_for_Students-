```python
dupl = df.duplicated().sum()
if dupl > 0:
    df = df.drop_duplicates().reset_index(drop = True)

df.shape
```

```python
df['CHAS'] = df['CHAS'].astype(str)
```

```python
df['RAD'] = df['RAD'].astype(str)
```

```python
df.info()
```

```python
cat_col = df.select_dtypes(include = 'object').columns
```

```python
cat_col
```

```python
num_col = df.select_dtypes(exclude = 'object').columns
```

```python
num_col
```

```python
for col in num_col:
    z_scores = (df[col] - df[col].mean()) / df[col].std()

    z_counts = (z_scores.abs() > 3).sum()

    print(f'Column {col}: {z_counts} outliers (Z-score > 3)')
```

#Data Analysis with Pandas

```python
corr = df.corr(method = 'spearman').round(3)
corr
```

```python
medv_corr = corr['MEDV'].drop('MEDV').sort_values(key = abs, ascending = False)
medv_corr
```

#FEATURE ENGINEERING

```python
df['TAX_PER_ROOM'] = df['TAX'] / df['RM']
df
```

```python
df['CRIM_PER_ZN'] = df['CRIM'] / (df['ZN'] + 1)
df
```

```python
import numpy as np

df['LOG_CRIM'] = np.log1p(df['CRIM'])
df
```

```python
df['LOG_ZN'] = np.log1p(df['ZN'])
df
```

```python
df['HIGH CRIM'] = (df['CRIM'] > df['CRIM'].quantile(0.75)).astype(int)
df[df['HIGH CRIM'] > 0]
```

```python
new_corr = df.corr(method = 'spearman').round(3)
```
